### broadcasting in numpy
- Broadcasting is a feature in NumPy that allows arrays of different shapes to perform arithmetic operations without explicitly copying data.

- It automatically "stretches" the smaller array to match the shape of the larger array.


```python
# broadcasting
a = np.arange(6).reshape(2,3)
b = np.arange(6).reshape(2,3)

print(a)
print(b)
print(a+b)
```

```python
# diff shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)

print(a)
print(b)
print(a+b)
```

### Broadcasting Rules

- NumPy compares array shapes from right to left.

- Two dimensions are compatible if:

1. They are equal, or
2. One of them is 1.

Otherwise, broadcasting fails.


- 1. Make the two arrays have the same number of dimensions.
If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.


2. Make each dimension of the two arrays the same size.
If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

### sigmoid 
A sigmoid function is a mathematical curve shaped like the letter S. It takes any real number and squashes it to a value between 0 and 1. The standard formula is . It is widely used in machine learning for logistic regression and neural networks to turn numbers into probabilities.

```python
# sigmoid
def sigmoid (array):
    return 1/(1 + np.exp ( - (array)))
a = np.arange(12)
sigmoid(a)
```

### Mean squsre error 
Mean Squared Error (MSE) is a number that tells you how close a model's guesses are to the real data. It finds the difference between each real value and guessed value, squares each difference so negative numbers do not cancel positive ones, and takes the average.


```python
# mean square error 
actual = np.random.randint(1,50,25)
pridicted = np.random.randint(1,50,25)
 
 
```

```python
# categorical cross entropy 
pridicted
```

```python
def mse(actual, pridicted):
    np.mean((actual - pridicted)**2)

mse(actual, pridicted)
```

### binary cross entropy
BCE=−N1​∑[ylog(y^​)+(1−y)log(1−y^​)]

```python
# binary cross entropy
actual = np.random.randint(1,50,25)
pridicted = np.random.randint(1,50,25)
```

```python
def binary_cross_entropy(actual, predicted):
    return -np.mean(
        actual * np.log(predicted) +
        (1 - actual) * np.log(1 - predicted)
    )

binary_cross_entropy(actual, pridicted)
```
